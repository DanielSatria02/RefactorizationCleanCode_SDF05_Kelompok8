```python
def orderTotal(individualPrice, quantity, isMember, shippingFee):
    subTotal = individualPrice * quantity

    if quantity > 100:
        discountedTotal = subTotal * 0.90
    else:
        discountedTotal = subTotal * 0.95

    if isMember:
        shippingCost = 0
    else:
        shippingCost = shippingFee

    return discountedTotal + shippingCost


individualPrice = float(input("Individual price: "))
quantity = int(input("Quantity: "))
isMember = input("Is the customer a member? (yes/no): ").strip().lower() == "yes"
shippingFee = float(input("Shipping fee: "))

print("Order total:", orderTotal(individualPrice, quantity, isMember, shippingFee))
```
